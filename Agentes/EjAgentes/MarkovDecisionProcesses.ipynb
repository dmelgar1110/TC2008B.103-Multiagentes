# Cadena de Markov



```python
import numpy as np
np.set_printoptions(precision=2, suppress=True)

# Estados definidos
S = ['S', 'N', 'L']  # S: Soleado, N: Nublado, L: Lluvioso

# Descripcion
description = {'S': 'Soleado', 'N': 'Nublado', 'L': 'Lluvioso'}

# Probabilidades de transición
P = np.array([
    [0.6, 0.3, 0.1],
    [0.4, 0.4, 0.2],
    [0.2, 0.3, 0.5]
])

# Probabilidad de transicion en n pasos
def prob_transicion(inicial, final, pasos = 1):
    Pn = np.linalg.matrix_power(P, pasos)
    i, j = S.index(inicial), S.index(final)
    return Pn[i, j]

# Ejemplos
inicial, final = 'S', 'S'
pasos = 2
p = prob_transicion(inicial, final, pasos)
print('Prob de transición de {} a {} en {} pasos: {:.2f}'.format(description[inicial], description[final], pasos, p))
```

# Proceso de Decisión de Markov (simple)

```python
# Posibles acciones
A = ['C', 'Q']  # C: Caminar, Q: Quedarse

# Recompensas (R)
recompensas = {
    'S': {'C': 10, 'Q': 2},
    'N': {'C': 5, 'Q': 3},
    'L': {'C': -10, 'Q': 1}
}

def recompensa_esperada(estado, accion):
    """
    Calcular la recompensa esperada para un estado y acción dados.
    """
    prob = P[S.index(estado)]
    esperada = sum(p * recompensas[nuevo_estado][accion] for p, nuevo_estado in zip(prob, S))
    return esperada


def elegir_accion(estado):
    """
    Elegir la acción con la mayor recompensa esperada.
    """
    recompensas_accion = {accion: recompensa_esperada(estado, accion) for accion in A}
    mejor_accion = max(recompensas_accion, key=recompensas_accion.get)
    return mejor_accion, recompensas_accion[mejor_accion]


# Ejemplo de uso
estado_actual = 'L'  # Comenzando desde un día nublado
rc = recompensa_esperada(estado_actual, 'C')
rq = recompensa_esperada(estado_actual, 'Q')
print('Recompensas esperadas para el estado: {} - C: {:.2f}, Q: {:.2f}'.format(estado_actual, rc, rq))

# Decisión
d = elegir_accion(estado_actual)
print('Opción elegida: {}'.format(d[0]))
```

# Recompensa esperada como producto matricial

```python
R = np.array([[10, 5, -10], [2, 3, 1]])
PR = np.matmul(P, np.transpose(R))

print('Recompensas esperadas para el estado: {} - C: {}, Q: {}'.format(estado_actual, PR[S.index(estado_actual),0], PR[S.index(estado_actual),1]))
```

# Proceso de Decisión de Markov (n pasos)


```python
def recompensa_esperada(estado, acciones):
    exp_reward = 0
    steps = 1
    for accion in acciones:
        PR = np.matmul(np.linalg.matrix_power(P, steps), np.transpose(R))
        exp_reward += PR[S.index(estado), A.index(accion)]
        steps += 1

    return exp_reward
```
